```python
print(f"Before: \t{len(events)}")
events = filters.order_objects_on_pt(events=events)
events = filters.jet_pt_eta_mask(events=events, min_pt=20, max_abs_eta=4.5)
events = filters.has_at_least_two_jets(events=events)
print(f"After 2jet: \t{len(events)}")

events = filters.has_atleast_1_muon(events)
print(f"After muon: \t{len(events)}")
events = filters.has_atleast_1_electron(events)
print(f"After elec: \t{len(events)}")
events = filters.leptons_have_opposite_charge(events)
print(f"After charge: \t{len(events)}")
events = filters.lepton_pt_mask(events, leading_pt=22, subleading_pt=15)
print(f"After lep pt: \t{len(events)}")
events = filters.lepton_eta_mask(events, electron_eta=2.5, muon_eta=2.5)
print(f"After lep eta: \t{len(events)}")
events = filters.leptons_deltaR_mask(events, deltaR=0.1)
print(f"After lep dR: \t{len(events)}")
events = filters.lepton_masses_mask(events, mass_cutoff=10)
print(f"After lep m: \t{len(events)}")
events = filters.lepton_jet_deltaR_mask(events, deltaR_thres=0.4) ##!!!!!!!
print(f"After l-j dR: \t{len(events)}")
events = filters.remove_btag(events, pt_thres=20, eta_thres=2.5)
print(f"After bjet: \t{len(events)}")


events = filters.central_jet_veto(events=events, pt_cutoff=20)
print(f"After CJV: \t{len(events)}")

events = filters.outside_lepton_veto(events)
print(f"After OLV: \t{len(events)}")

events = filters.jets_mass_mask(events, mass_cutoff=450)
print(f"After jet m: \t{len(events)}")
events = filters.jet_rapidity_mask(events, rapidity_diff_cutoff=2.1)
print(f"After jet rap: \t{len(events)}")
events = filters.lepton_phi_mask(events, delta_phi_cutoff=1.4)
print(f"After lep phi: \t{len(events)}")
```
